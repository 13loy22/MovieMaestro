# Data Generation Code

This section generates synthetic data for a movie recommendation system. The data is created in a sequential manner to ensure dependencies are met:

1.  **`movies.csv`**: Contains information about movies (ID, rating, votes, runtime, release year, revenue, genre).
2.  **`user_data.json`**: Contains user profiles (name, age, preferred genres).
3.  **`viewing_history.csv`**: Contains records of which users watched which movies and their ratings.

```python
import numpy as np
import pandas as pd

# Number of movies
number_of_movies = 1000

# Generate movie data
movie_ids = np.arange(1, number_of_movies + 1)
ratings = np.random.uniform(1, 10, number_of_movies)
votes = np.random.randint(100, 100000, number_of_movies)
runtime = np.random.randint(80, 181, number_of_movies)
release_year = np.random.randint(2000, 2026, number_of_movies)
revenue = np.random.uniform(1, 500, number_of_movies)
genres = np.random.choice(
    ["Action", "Comedy", "Drama", "Horror", "Sci-Fi"],
    number_of_movies
)

# Create a DataFrame
df = pd.DataFrame({
    'movie_id': movie_ids,
    'rating': ratings,
    'votes': votes,
    'runtime': runtime,
    'release_year': release_year,
    'revenue': revenue,
    'genre': genres
})

# Save to CSV
df.to_csv('movies.csv', index=False)

print("Data generated and saved to movies.csv")
display(df.head())
```

```python
import json
import random

# Number of users to generate
number_of_users = 1000

# Aligning with the genres from your movie generation code
genres = ["Action", "Comedy", "Drama", "Horror", "Sci-Fi"]

# A list of sample names to draw from
first_names = [
    "John", "Emma", "Michael", "Sophia", "William", "Olivia", "James", "Ava",
    "Alexander", "Isabella", "Daniel", "Mia", "Matthew", "Charlotte", "Joseph",
    "Amelia", "David", "Harper", "Jackson", "Evelyn", "Lucas", "Abigail"
]

user_data = []

for i in range(number_of_users):
    # 1. Generate basic demographics
    user_id = i + 1  # Assign a unique user ID starting from 1
    name = random.choice(first_names)
    age = random.randint(18, 65)

    # 2. Generate preferences (1 to 3 random genres)
    num_preferences = random.randint(1, 3)
    preferences = random.sample(genres, num_preferences)

    # 3. Construct the user dictionary
    user = {
        "user_id": user_id,
        "name": name,
        "age": age,
        "preferences": preferences
    }

    user_data.append(user)

# Print the first 2 users to the console to verify the format
print(json.dumps(user_data[:2], indent=4))

# Export all 1000 users to a JSON file
with open("user_data.json", "w") as file:
    json.dump(user_data, file, indent=4)

print(f"\nSuccessfully generated {number_of_users} users and saved to 'user_data.json'.")
```

```python
import json
import random
import pandas as pd

# -----------------------------
# STEP 3: GENERATE VIEWING HISTORY
# -----------------------------

# Load users generated in Step 2
with open("user_data.json", "r") as file:
    users = json.load(file)

# Load movies generated in Step 1
movies = pd.read_csv("movies.csv")

# Get the movie IDs from the actual movie dataset
movie_ids = movies["movie_id"].tolist()

# List to store all viewing events
viewing_history = []

# Generate viewing history for every user
for user_id, user in enumerate(users, start=1):

    # Each user watches between 5 and 20 movies
    number_of_views = random.randint(5, 20)

    # Select different movies for this user
    selected_movies = random.sample(
        movie_ids,
        min(number_of_views, len(movie_ids))
    )

    for movie_id in selected_movies:

        # Generate a rating from 1 to 5
        rating = random.randint(1, 5)

        # Create one viewing event
        viewing_history.append({
            "user_id": user_id,
            "movie_id": movie_id,
            "rating": rating
        })

# Convert to DataFrame
viewing_history_df = pd.DataFrame(viewing_history)

# Save as CSV
viewing_history_df.to_csv(
    "viewing_history.csv",
    index=False
)

# Show the first 10 records
print("\nFirst 10 viewing history records:")
print(viewing_history_df.head(10))

# Show statistics
print("\nTotal viewing events:", len(viewing_history_df))
print("Total users:", viewing_history_df["user_id"].nunique())
print("Total movies watched:", viewing_history_df["movie_id"].nunique())

print("\nSuccessfully generated viewing_history.csv")
```
